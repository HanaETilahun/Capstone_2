### Question 1: Territory Managers, Store IDs, and Cities

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
store_details1 = pd.read_csv("StoreDetail.csv")

store_details1.head()
```

```python
south_stores = store_details1[store_details1["Region"] == "South"]

south_stores[["Store ID", "Store Location", "State", "Territory Manager"]]
```

#### Question 1 Answer

### Question 2: Monthly Total Revenue for In-Store Sales

The next analysis looks at monthly total revenue for in-store sales in the South territory over the full time period covered in the dataset. To answer this question, I merged the sales data with the store detail data and grouped the results by month.

```python
# Load store sales data
store_sales = pd.read_csv("storeDetails.csv")

store_sales.head()
```

```python
# Merge sales data with store detail data
sales_merged = pd.merge(
    store_sales,
    store_details1,
    on="Store ID"
)

sales_merged.head()
```

```python
# Convert transaction date column to datetime
sales_merged["Transaction Date"] = pd.to_datetime(
    sales_merged["Transaction Date"]
)

# Create month column
sales_merged["Month"] = sales_merged["Transaction Date"].dt.to_period("M")

sales_merged.head()
```

```python
# Filter South region sales
south_sales = sales_merged[sales_merged["Region"] == "South"]

south_sales.head()
```

```python
# Calculate monthly revenue
monthly_revenue = south_sales.groupby("Month")["Sale Amount"].sum()

monthly_revenue
```

```python
monthly_revenue.plot(
    kind="bar",
    figsize=(10,6)
)

plt.title("Monthly Revenue - South Territory")
plt.xlabel("Month")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)

plt.show()
```

#### Question 2 Answer

The chart above shows monthly total revenue for in-store sales in the South territory. The analysis helps identify sales trends, periods of stronger revenue performance, and opportunities for future marketing focus.

### Question 3: Store Sales Performance Ranking

This analysis ranks store performance in the South territory based on total sales revenue. The goal is to identify the top-performing stores and compare overall sales performance across store locations.

```python
# Calculate total sales revenue by store
store_performance = south_sales.groupby(
    ["Store ID", "Store Location"]
)["Sale Amount"].sum().reset_index()

store_performance.head()
```

```python
# Sort stores from highest to lowest sales revenue
store_performance = store_performance.sort_values(
    by="Sale Amount",
    ascending=False
)

store_performance
```

```python
# Display top 5 performing stores
top_5_stores = store_performance.head(5)

top_5_stores
```

```python
top_5_stores.plot(
    x="Store Location",
    y="Sale Amount",
    kind="bar",
    figsize=(10,6)
)

plt.title("Top 5 Performing Stores")
plt.xlabel("Store Location")
plt.ylabel("Total Sales Revenue")
plt.xticks(rotation=45)

plt.show()
```

#### Question 3 Answer

The chart above ranks the top-performing stores in the South territory based on total sales revenue. Stores with the highest revenue demonstrate stronger sales performance and customer activity compared to other locations.

### Question 4: Top Customers in the South Territory

This analysis identifies the top customers in the South territory by comparing the RewardsID from the sales data with the cust_id from the customer list data. Customers are ranked based on their total purchase amount.

```python
# Load customer list data
customer_list = pd.read_csv(
    "customer_List.csv",
    sep="|"
)

customer_list.head()
```

```python
# Merge customer data with South sales data
customer_sales = pd.merge(
    south_sales,
    customer_list,
    left_on="RewardsID",
    right_on="cust_id"
)

customer_sales.head()
```

```python
# Calculate total spending by customer
top_customers = customer_sales.groupby(
    ["cust_id", "name"]
)["Sale Amount"].sum().reset_index()

top_customers.head()
```

```python
# Rank customers by total spending
top_customers = top_customers.sort_values(
    by="Sale Amount",
    ascending=False
)

top_customers.head(10)
```

#### Question 4 Answer

The table above shows the top customers in the South territory ranked by total purchase amount. These customers contribute the highest sales revenue and may represent valuable targets for loyalty programs and future marketing campaigns.

### Question 5: Transactions and Revenue by Product Category

This analysis examines the number of transactions and total sales revenue per month by product category in the South territory. The goal is to identify the most popular product categories and potential opportunities for future business growth.

```python
# Load product category data
product_categories = pd.read_csv("ProductCategories.csv")

product_categories.head()
```

```python
# Load product data
products = pd.read_csv("Products.csv")

products.head()
```

```python
# Merge products with product categories
products_merged = pd.merge(
    products,
    product_categories,
    on="CategoryID"
)

products_merged.head()
```

```python
# Merge sales data with product information
sales_products = pd.merge(
    south_sales,
    products_merged,
    on="Prod Num"
)

sales_products.head()
```

```python
# Count transactions per month by category
monthly_transactions = sales_products.groupby(
    ["Month", "Category"]
)["Sale Amount"].count().reset_index()

monthly_transactions
```

```python
# Calculate monthly revenue by category
monthly_category_revenue = sales_products.groupby(
    ["Month", "Category"]
)["Sale Amount"].sum().reset_index()

monthly_category_revenue
```

```python
# Top product categories by revenue
top_categories = monthly_category_revenue.sort_values(
    by="Sale Amount",
    ascending=False
)

top_categories.head(10).plot(
    x="Category",
    y="Sale Amount",
    kind="bar",
    figsize=(10,6)
)

plt.title("Top Product Categories by Revenue")
plt.xlabel("Product Category")
plt.ylabel("Revenue")
plt.xticks(rotation=45)

plt.show()
```

#### Question 5 Answer

The analysis shows the number of transactions and total sales revenue by product category in the South territory. Categories with the highest revenue and transaction counts appear to be the most popular among customers. Lower-performing categories may provide opportunities for additional marketing campaigns, promotions, or product expansion strategies.

Based on the analysis, marketing efforts next quarter should focus on the highest-performing product categories and store locations in the South territory.
